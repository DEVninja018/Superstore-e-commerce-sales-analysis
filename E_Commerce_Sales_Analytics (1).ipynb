```python
# Import necessary libraries like pandas,matplotlib,seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
```

```python
# Load the 'Sample - Superstore.csv' dataset
data=pd.read_csv("Sample - Superstore.csv",encoding="latin1")
data.head()
```

```python
# Generate descriptive statistics for numerical columns.
data.describe()
```

```python
# Display a concise summary of the DataFrame
data.info()
```

```python
# Converting 'Order Date' and 'Ship Date' columns from object (string) type to datetime objects.
data['Order Date']=pd.to_datetime(data['Order Date'])
data['Ship Date']=pd.to_datetime(data['Ship Date'])
```

```python
# Re-display the DataFrame.'Order Date' and 'Ship Date' are now of datetime type.
data.info()
```

```python
# Extracting the month, year, and day of the week from the 'Order Date' column.
data['Order Month']=data['Order Date'].dt.month
data['Order Year']=data['Order Date'].dt.year
data['Order Day of week']=data['Order Date'].dt.dayofweek
```

```python
#verification of the new columns 'Order Month', 'Order Year', and 'Order Day of week.
data
```

```python
# Group the data by 'Order Month' and calculate the sum of 'Sales' for each month.
sales_by_month=data.groupby('Order Month')['Sales'].sum().reset_index()
sales_by_month
```

```python
#line plot to visualize the total sales trend over different months.
figure=plt.figure(figsize=(10,10))

sns.lineplot(data=sales_by_month,x='Order Month',y='Sales')

plt.xlabel('Order Month')
plt.ylabel('Sales')
plt.title('Sales by Month')

plt.show()
```

```python
# Grouped the data by 'Category' and calculate the sum of 'Sales' for each category.
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
# Created a bar plot to visualize total sales across different product categories.
figure=plt.figure(figsize=(20,10))

sns.barplot(data=sales_by_category,x='Category',y='Sales',palette="muted")

plt.xlabel('Category')
plt.ylabel('Sales')
plt.title('Sales by Category')

plt.show()
```

```python
# Grouped the data by 'Sub-Category' and calculate the sum of 'Sales' for each sub-category.
sales_by_subcategory=data.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_subcategory
```

```python
# Created a bar plot to visualize total sales across different product sub-categories.
figure=plt.figure(figsize=(25,15))

sns.barplot(data=sales_by_subcategory,x='Sub-Category',y='Sales',palette="colorblind")

plt.xlabel('Sub-Category')
plt.ylabel('Sales')
plt.title('Sales by Sub-Category')

plt.show()
```

```python
# Grouped the data by 'Order Month' and calculate the sum of 'Profit' for each month.
profit_by_month=data.groupby('Order Month')['Profit'].sum().reset_index()
profit_by_month
```

```python
# Created a line plot to visualize the total profit trend over different months.
figure=plt.figure(figsize=(10,10))

sns.lineplot(data=profit_by_month,x='Order Month',y='Profit')

plt.xlabel('Order Month')
plt.ylabel('Profit')
plt.title('Profit by Month')

plt.show()
```

```python
# Grouped the data by 'Category' and calculate the sum of 'Profit' for each category.
profit_by_category=data.groupby('Category')['Profit'].sum().reset_index()
profit_by_category
```

```python
# Created a pie chart to visualize the distribution of profit across different product categories.

figure=plt.figure(figsize=(20,10))

plt.pie(profit_by_category['Profit'],labels=profit_by_category['Category'],autopct='%1.1f%%',)

plt.title('Profit by Category')

plt.show()
```

```python
# Grouped the data by 'Sub-Category' and calculate the sum of the absolute 'Profit' for each sub-category.`abs()` helps to visualize the magnitude of profit/loss without considering the direction.
profit_by_subcategory=data.groupby('Sub-Category')['Profit'].sum().abs().reset_index()
profit_by_subcategory
```

```python
# Created a bar chart to visualize the absolute profit across different product sub-categories.
figure=plt.figure(figsize=(25,15))

plt.bar(profit_by_subcategory['Sub-Category'],profit_by_subcategory['Profit'])

plt.xlabel('Sub-Category')
plt.ylabel('Profit')
plt.title('Profit by Sub-Category')

plt.show()
```

```python
# Aggregated 'Sales' and 'Profit' by 'Segment' to analyze customer segment performance.
segment_performance=data.groupby('Segment').aggregate({'Sales':'sum','Profit':'sum'}).reset_index()
```

```python
# Created two bar plots to visualize 'Sales' and 'Profit' for each customer segment.
plt.subplot(2,1,1)
sns.barplot(data=segment_performance,x='Segment',y='Sales')
plt.xlabel('Segment')
plt.ylabel('Sales')
plt.title("Sales by Customer Segment")

plt.subplot(2,1,2)
sns.barplot(data=segment_performance,x='Segment',y='Profit')
plt.xlabel('Segment')
plt.ylabel('Profit')
plt.title("Profit by Customer Segment")

plt.subplots_adjust(hspace=0.5)
plt.show()
```

```python
# Re-aggregated 'Sales' and 'Profit' by 'Segment' to prepare for calculating the sales-to-profit ratio.
sale_to_profit_ratio=data.groupby('Segment').aggregate({'Sales':'sum','Profit':'sum'}).reset_index()
```

```python
# Calculated the 'Sales to Profit ratio' for each segment by dividing total sales by total profit.
sale_to_profit_ratio['Sales to Profit ratio']=sale_to_profit_ratio['Sales']/sale_to_profit_ratio['Profit']
print(sale_to_profit_ratio[['Segment','Sales to Profit ratio']])
```

*Insight: Corporate and Home Office segments appear to have slightly better sales to profit ratios compared to Consumer, suggesting better efficiency in converting sales to profit.*


```python

correlation = data['Sales'].corr(data['Profit'])
print(f"Correlation between Sales and Profit: {correlation}")
```

***A correlation of 0.479 suggests a moderate positive linear relationship between Sales and Profit, indicating if Sales increases/decreases,Profit also increases/decreases***
